```python
import pandas as pd
data = {
    "car":["honda city","hyundai creta","maruti swift","tat nexon","toyota innova","honda amaze","kia seltos","maruti baleno"],
    "company":["honda","hyundai","maruti","tata","toyota","honda","kia","maruti"],
    "year":[2022,2023,2021,2023,2022,2020,2024,2021],
    "price":[1200000,1500000,700000,1100000,2200000,650000,1600000,800000],
    "mileage":[18,17,23,19,14,20,16,22],
    "Fuel":["petrol","disel","petrol","disel","disel","petrol","petrol","petrol"],
    "rating":[4.5,4.3,4.2,4.4,4.6,4.0,4.5,4.1]
}
df=pd.DataFrame(data)
```
