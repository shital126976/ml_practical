```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

#convert horsepower to numeric,replacing any non-numeric values (like'?') with NaN
df['horsepower']=pd.to_numeric(df['horsepower'],errors='coerce')
#fill any missing alues in horsepower with the median value
df['horsepower']=df['horsepower'].fillna(df['horsepower'].median())

# select existing numeric columns from your dataset
features = ['wheel-base', 'length', 'width', 'height', 'curb-weight', 'engine-size', 'compression-ratio', 'horsepower', 'city-mpg', 'highway-mpg']
X = df[features]

#rescale data
scaler=MinMaxScaler()
rescaled_data=scaler.fit_transform(X)

#create a new dataframe with the rescaled data
df_rescaled=pd.DataFrame(rescaled_data,columns=X.columns)
#display result
print(df_rescaled.head())
```

```python
from sklearn.preprocessing import LabelEncoder

# create LabelEncoder object
encoder = LabelEncoder()

# Encode categorical column
df['make'] = encoder.fit_transform(df['make'])

# Display result
print(df[['make']].head())
```

```python
mport pandas as pd
import matplotlib.pyplot as plt
from sklearin.preprocessing import StandardScaler
from sklearn.decomposition import PCA

#Select numeric features
X=df[['city-mpg', 'highway-mpg', 'curb-weight', 'engine-size', 'horsepower']]

#Handle missing values
X=X.fillna(X.mean())

#Standardize the data
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)

#Perform PCA
pca=PCA(n_components=2)
principle_components=pca.fit_transform(X_scaled)

#Plot
plt.scatter(principle_components[:,0],principle_components[:,1])
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Automobile Data")
plt.show()

```
